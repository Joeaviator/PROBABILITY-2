```python
#Sets
S1 = {"A", "B", "C", "D", "E"}
S2 = {"B", "D", "V", "X", "Y", "Z"}
#union
union_set=S1.union(S2)
print("Final List Of Friends: ", union_set)
```

```python
#Sets
S1 = {"A", "B", "C", "D", "E"}
S2 = {"B", "D", "V", "X", "Y", "Z"}
#Intersecion
intersection_set=S1.intersection(S2)
print("Common invited Frineds: ", intersection_set)
```

```python
dice={1,2,3,4,5,6}#Possibleevents
#define events
event_A={2,4,6}#Even number
event_B={3,4,5,6}#More than 2
probaility_A=len(event_A)/len(dice)
probaility_B=len(event_B)/len(dice)
print(f"Probaility of A is : {probaility_A:.2f}")
print(f"Probability of B is : {probaility_B:.2f}")
both_p=probaility_A*probaility_B
print(f"Probaility of Both Events happening: {both_p:.2f}")
```

```python
#Addition rule
intersection_a=event_A.intersection(event_B)
intersection_prob=len(intersection_a)/len(dice)
result=probaility_A+probaility_B-intersection_prob
print(f"{result:.2f}")
```
